```python
# Импортируем ключевой класс YOLO из библиотеки
from ultralytics import YOLO
# Используем библиотеку cv2 для визуализации
import cv2

# Загружаем предобученную модель
# Если файла 'yolov8n.pt' нет в папке, Python скачает его автоматически
print("Загрузка модели...")
model = YOLO('yolov8n.pt') 
print("Модель загружена.")

# Источник данных (может быть ссылкой или путем к локальному файлу)
source = 'https://ultralytics.com/images/bus.jpg'

# Выполняем инференс (предсказание)
# Метод .predict() возвращает список объектов Results
results = model.predict(source)

# Получаем результат для нашего единственного изображения
result = results[0]

# Метод .plot() автоматически рисует рамки, подписи классов и уверенность.
# Результат возвращается в формате массива NumPy (цветовое пространство BGR)
annotated_image = result.plot()

# Отображаем результат в отдельном окне
cv2.imshow("YOLOv8 Detection Results", annotated_image)

# Ждём нажатия любой клавиши, чтобы закрыть окно и завершить работу скрипта
cv2.waitKey(0) 
cv2.destroyAllWindows() 
```
